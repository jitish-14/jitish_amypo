```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)




data = pd.read_excel(r"C:\Users\jitis\Downloads\Student_Performance_Dataset (2).xlsx")

print("STUDENT DATASET")
print(data)



print("\nFirst 5 Rows:")
print(data.head())

print("\nDataset Information:")
data.info()

print("\nMissing Values:")
print(data.isnull().sum())

print("\nStatistical Summary:")
print(data.describe())



X = data[["Study Hours", "Attendance (%)", "Previous Marks"]]


y = data["Final Marks"]


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

print("\nTraining Data:", X_train.shape)
print("Testing Data:", X_test.shape)


reg_model = LinearRegression()


reg_model.fit(X_train, y_train)

print("\nLinear Regression Model Trained Successfully")


y_pred = reg_model.predict(X_test)


regression_result = pd.DataFrame({
    "Actual Marks": y_test,
    "Predicted Marks": y_pred
})

print("\nActual vs Predicted Marks:")
print(regression_result)


mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print("\nRegression Evaluation Metrics:")
print("MAE:", round(mae, 2))
print("MSE:", round(mse, 2))
print("RMSE:", round(rmse, 2))
print("R2 Score:", round(r2, 2))




new_student = pd.DataFrame({
    "Study Hours": [4],
    "Attendance (%)": [75],
    "Previous Marks": [60]
})

predicted_marks = reg_model.predict(new_student)

print("\nNew Student Details:")
print(new_student)

print("\nPredicted Final Marks:", round(predicted_marks[0], 2))



data["Result_Label"] = data["Result"].map({
    "Fail": 0,
    "Pass": 1
})

X_class = data[["Study Hours", "Attendance (%)", "Previous Marks"]]


y_class = data["Result_Label"]

print("\nTarget Values:")
print(y_class)


X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
    X_class,
    y_class,
    test_size=0.25,
    random_state=42,
    stratify=y_class
)

print("\nTraining Data:", X_train_class.shape)
print("Testing Data:", X_test_class.shape)


class_model = LogisticRegression()


class_model.fit(X_train_class, y_train_class)

print("\nLogistic Regression Model Trained Successfully")


y_class_pred = class_model.predict(X_test_class)


predicted_results = []

for value in y_class_pred:
    if value == 1:
        predicted_results.append("Pass")
    else:
        predicted_results.append("Fail")


classification_result = pd.DataFrame({
    "Actual Result": y_test_class.map({0: "Fail", 1: "Pass"}),
    "Predicted Result": predicted_results
})

print("\nActual vs Predicted Results:")
print(classification_result)



accuracy = accuracy_score(y_test_class, y_class_pred)

precision = precision_score(
    y_test_class,
    y_class_pred,
    zero_division=0
)

recall = recall_score(
    y_test_class,
    y_class_pred,
    zero_division=0
)

f1 = f1_score(
    y_test_class,
    y_class_pred,
    zero_division=0
)

print("\nClassification Evaluation Metrics:")
print("Accuracy:", round(accuracy, 2))
print("Precision:", round(precision, 2))
print("Recall:", round(recall, 2))
print("F1 Score:", round(f1, 2))


print("\nConfusion Matrix:")
print(confusion_matrix(y_test_class, y_class_pred))


print("\nClassification Report:")
print(classification_report(
    y_test_class,
    y_class_pred,
    target_names=["Fail", "Pass"],
    zero_division=0
))




predicted_class = class_model.predict(new_student)

if predicted_class[0] == 1:
    final_result = "Pass"
else:
    final_result = "Fail"

print("\nPredicted Student Result:", final_result)




comparison = pd.DataFrame({
    "Comparison": [
        "Algorithm Used",
        "Target Variable",
        "Type of Prediction",
        "Evaluation Metrics",
        "Real-World Application"
    ],

    "Regression": [
        "Linear Regression",
        "Final Marks",
        "Numerical Prediction",
        "MAE, MSE, RMSE, R2 Score",
        "Predict student examination marks"
    ],

    "Classification": [
        "Logistic Regression",
        "Pass/Fail Result",
        "Categorical Prediction",
        "Accuracy, Precision, Recall, F1 Score",
        "Identify students at risk of failing"
    ]
})

print(comparison.to_string(index=False))






print("Study Hours:", new_student["Study Hours"].iloc[0])
print("Attendance:", new_student["Attendance (%)"].iloc[0], "%")
print("Previous Marks:", new_student["Previous Marks"].iloc[0])

print("Predicted Final Marks:", round(predicted_marks[0], 2))
print("Predicted Pass/Fail Result:", final_result)

print("\nProgram Completed Successfully!")
```
